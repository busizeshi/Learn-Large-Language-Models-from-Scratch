### 范数

范数（norm）是 衡量一个向量“有多大”的函数 。注意它不只有一个定义，而是满足以下三条性质的一类函数，记作 ‖x‖：

1. 正定性 ：‖x‖ ≥ 0，且 ‖x‖ = 0 当且仅当 x = 0（零向量的“大小”才是 0）
2. 齐次性 ：‖αx‖ = |α|·‖x‖（向量放大几倍，大小也放大几倍）
3. 三角不等式 ：‖x + y‖ ≤ ‖x‖ + ‖y‖（两边之和大于等于第三边，和几何直觉一致）

**L2 范数 （欧几里得范数）——就是初中学的“距离”公式：**

‖x‖₂ = √(x₁² + x₂² + ... + xₙ²)


````python
u = torch.tensor([3.0, -4.0])
torch.linalg.norm(u)      # tensor(5.)  ← 3²+4²=25, √25=5
````

L1 范数 ——所有元素绝对值之和：

‖x‖₁ = |x₁| + |x₂| + ... + |xₙ|

````python
torch.abs(u).sum()        # tensor(7.)  ← 3 + 4
````

# 微积分

## 导数：瞬时变化率

f'(x) = lim(h→0) [f(x+h) - f(x)] / h

````python
x = torch.tensor(2.0, requires_grad=True)  # 要求 PyTorch 追踪 x 的梯度
y = x**3                                    # f(x) = x³
y.backward()                                # 自动求导
print(x.grad)                               # tensor(12.) ← 手算: 3x² = 3·2² = 12
````

## 偏导数

∇f(x) = (∂f/∂x₁, ∂f/∂x₂, ..., ∂f/∂xₙ)

````python
x = torch.tensor([1.0, -1.0], requires_grad=True)
f = (x**2).sum()     # f = x₁² + x₂²
f.backward()
print(x.grad)        # tensor([2., -2.]) ← 手算 (2x₁, 2x₂) = (2, -2) ✓
````

### 链式法则：复合函数求导

y = f(u), u = g(x)  →  dy/dx = (dy/du) · (du/dx)

例子：`y = (2x+1)³` 。设 u = 2x+1，则 dy/du = 3u²，du/dx = 2，所以：

dy/dx = 3u² · 2 = 6(2x+1)²

# 概率

简单地说，**机器学习就是做出预测**。

- 根据病人的临床病史，预测他下一年心脏病发作的*概率*
- 异常检测中，评估一组发动机读数属于正常运行的概率
- 推荐系统中，估计用户购买某本书的概率

概率是一种灵活的语言，用于说明我们的**确定程度**。

以猫狗分类为例：
- 完全肯定图像是猫：P(y="猫") = 1
- 毫无证据时：P(y="猫") = P(y="狗") = 0.5
- 不十分确定：0.5 < P(y="猫") < 1

## 基本概率论

假设我们掷骰子，想知道看到 1 的几率。如果骰子是公平的，六个结果 {1,…,6} 等可能发生，因此 1 出现的概率为 1/6。

但现实中的骰子可能有瑕疵，检查的唯一方法是**多次投掷并记录结果**：

- 把某个值出现的次数 ÷ 总投掷次数，作为该事件概率的**估计值**
- **大数定律**：随着投掷次数增加，估计值会越来越接近真实的潜在概率

````python
%matplotlib inline
import random
import torch
from torch.distributions import multinomial
import matplotlib.pyplot as plt
````

在统计学中，把从概率分布中抽取样本的过程称为**抽样**（sampling）。

- **分布**（distribution）：对事件的概率分配
- **多项分布**（multinomial distribution）：把概率分配给一些离散选择的分布

掷一次骰子 = 抽一个样本，输出向量的索引 i 处的值是采样结果中 i+1 出现的次数：

````python
fair_probs = torch.ones([6]) / 6                # 公平骰子：每个面概率 1/6
multinomial.Multinomial(1, fair_probs).sample() # 掷一次
````

估计骰子公平性时，需要从同一分布生成**多个样本**。用框架的函数一次抽取多个样本（相当于掷 10 次骰子），比 Python 的 for 循环快得多：

````python
multinomial.Multinomial(10, fair_probs).sample()   # 掷 10 次，统计每个面出现的次数
````

模拟 1000 次投掷，统计每个数字被投中的次数，用**相对频率**作为真实概率的估计：

````python
counts = multinomial.Multinomial(1000, fair_probs).sample()
counts / 1000   # 相对频率，每个值都应接近 1/6 ≈ 0.167
````

观察估计概率如何随数据增多而**收敛**到真实概率：进行 500 组实验，每组抽取 10 个样本（投 10 次）。

````python
counts = multinomial.Multinomial(10, fair_probs).sample((500,))
cum_counts = counts.cumsum(dim=0)                               # 沿实验组累加
estimates = cum_counts / cum_counts.sum(dim=1, keepdims=True)   # 累计相对频率

plt.figure(figsize=(6, 4.5))
for i in range(6):
    plt.plot(estimates[:, i].numpy(), label=("P(die=" + str(i + 1) + ")"))
plt.axhline(y=0.167, color='black', linestyle='dashed')   # 真实概率 1/6
plt.xlabel('Groups of experiments')
plt.ylabel('Estimated probability')
plt.legend();
````

每条实线对应骰子的一个面，给出该面在每组实验后出现的估计概率。随着实验组数（数据量）增加，6 条曲线都向真实概率 1/6 ≈ 0.167 收敛——这就是**大数定律**。

### 概率论公理

- **样本空间**（sample space）：所有可能结果的集合，如掷骰子的 S = {1,2,3,4,5,6}，其中每个元素是一个**结果**（outcome）
- **事件**（event）：给定样本空间中的一组随机结果，如"看到奇数" = {1,3,5}。若投出 3 点，因为 3 ∈ {1,3,5}，就称"看到奇数"这一事件发生了

概率是把集合映射到真实值的函数 P(A)，满足以下公理（科尔莫戈罗夫，1933）：

1. 对任意事件 A，概率非负：P(A) ≥ 0
2. 整个样本空间的概率为 1：P(S) = 1
3. 对**互斥**事件 A₁, A₂, …（任意 i≠j 都有 Aᵢ∩Aⱼ=∅）具有可加性：
   P(A₁∪A₂∪…) = P(A₁) + P(A₂) + …

有了这个公理系统，可以避开关于随机性的哲学争论，用数学语言严格推理。

### 随机变量

**随机变量**（random variable）几乎可以是任何数量，它在随机实验的一组可能性中取一个值。

- P(X = 5)：随机变量 X 取值为 5 的概率
- P(X)：随机变量 X 上的**分布**，即 X 取各个值的概率
- P(1 ≤ X ≤ 3)：X 从 {1,2,3} 中取值的概率

**离散** vs **连续**随机变量：
- 离散：取值可枚举（如骰子的每一面）
- 连续：如身高、体重。测量两人身高是否**完全相同**意义不大；问"身高恰好是 1.80 米"的概率为 0，但**密度**不为 0；有意义的是落在区间（如 1.79~1.81 米）的概率

## 处理多个随机变量

很多场景需要同时考虑多个随机变量：疾病与症状的关系、图像的数百万个像素 + 标签 + 元数据（位置、时间、光圈等），它们都是**联合发生**的随机变量。

### 联合概率

**联合概率** P(A=a, B=b)：A=a 和 B=b **同时**满足的概率。

P(A=a, B=b) ≤ P(A=a)：要同时发生，A=a 必须发生，所以联合概率不会大于单个事件的概率。

### 条件概率

由联合概率引出的比率——**条件概率** P(B=b | A=a)：在 A=a 已发生的前提下，B=b 的概率。

P(B=b | A=a) = P(A=a, B=b) / P(A=a)

### 贝叶斯定理

由乘法法则 P(A,B) = P(B|A)·P(A) 与对称性 P(A,B) = P(A|B)·P(B)，可得统计学中最有用的公式之一：

P(A|B) = P(B|A)·P(A) / P(B)

其中 P(A,B) 是**联合分布**，P(A|B) 是**条件分布**。它让我们能在"原因导致结果"与"结果反推原因"之间转换概率。

### 边际化

**求和法则**：B 的概率等于对 A 的所有可能取值，把联合概率聚合起来：

P(B) = Σ_A P(A, B)

这也称为**边际化**（marginalization），结果称为**边际概率** / **边际分布**。

### 独立性

如果 A 的发生与 B 无关，称 A 与 B **独立**（记作 A ⊥ B），此时：

P(A|B) = P(A)  ⟺  P(A, B) = P(A)·P(B)

（两个随机变量独立，当且仅当联合分布等于各自分布的乘积）

- 独立：连续两次掷骰子
- 依赖：灯开关的位置与房间亮度（可能灯泡坏了、电源故障）

**条件独立** A ⊥ B | C：给定 C 时，P(A,B|C) = P(A|C)·P(B|C)

### 应用：HIV 检测（贝叶斯定理实战）

假设一项 HIV 检测相当准确：

- 患者健康（H=0）但测试**误报**为阳性的概率只有 1%：P(D₁=1 | H=0) = 0.01
- 患者真感染（H=1）时，测试**绝不会漏报**：P(D₁=1 | H=1) = 1
- 疾病本身罕见，先验概率 P(H=1) = 0.0015

问题：一个人检测出阳性，他真的感染 HIV 的概率有多大？

````python
P_H = torch.tensor([1 - 0.0015, 0.0015])   # 先验: P(H=0 健康), P(H=1 感染)
P_D1_given_H = torch.tensor([0.01, 1.0])   # 似然: P(D1=1|H=0)=0.01 误报, P(D1=1|H=1)=1 必报

# 分母用边际化求 P(D1=1)，再代入贝叶斯定理
P_D1 = torch.dot(P_D1_given_H, P_H)
P_H_given_D1 = P_D1_given_H * P_H / P_D1
P_H_given_D1   # tensor([0.8693, 0.1306])
````

尽管检测非常准确，阳性结果下真正感染的概率只有约 **13.06%**！

原因：疾病极其罕见，健康人基数太大，1% 的误报产生的**假阳性**人数远多于真阳性。

**改进：再做一次独立检测 D₂。** 两次独立检测都误报的概率只有 0.01 × 0.01 = 0.0001：

````python
P_D2_and_D1_given_H = torch.tensor([0.0001, 1.0])   # 两次独立误报: 0.01 * 0.01
P_D1_and_D2 = torch.dot(P_D2_and_D1_given_H, P_H)
P_H_given_D1_and_D2 = P_D2_and_D1_given_H * P_H / P_D1_and_D2
P_H_given_D1_and_D2   # tensor([0.0624, 0.9376])
````

第二次检测把置信度从约 13.06% 提升到约 **93.76%**（错误率降到约 6.24%）。

核心结论：**条件概率的正确解读依赖先验**。即使检测手段很准，对极罕见的事件，单次检测的阳性结果也不足以下定论。

## 期望和方差

**期望**（expectation）：随机变量按概率加权的平均值，刻画分布的中心位置。

E[X] = Σ x · P(X = x)

**方差**（variance）：随机变量偏离期望的程度（偏差平方的期望）。

Var(X) = E[(X − μ)²] = E[X²] − μ²，其中 μ = E[X]

- 方差的平方根称为**标准差**（统计学家）或**均方根 RMS**（工程师）
- 方差越小，随机变量的取值越集中在期望附近

````python
# 以公平骰子为例：E[X] = 3.5，Var(X) = 35/12 ≈ 2.9167
x = torch.arange(1., 7.)           # 骰子的 6 个取值
p = torch.ones(6) / 6              # 每个取值的概率

mu = (p * x).sum()                 # 期望 E[X]
var = (p * x**2).sum() - mu**2     # 方差 = E[X^2] - μ^2
mu, var
````
