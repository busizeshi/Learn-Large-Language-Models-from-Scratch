# 机器学习的关键组件

## 数据 

每个数据集由一个个样本组成，大多数时候，它们遵循独立同分布。

通常每个样本由一组被称为特征或协变量的属性组成。监督学习中，要预测的一个属性，被称为标签或目标变量。

- 更多的数据可以被用于训练出更强大的模型，从而减少对预先假设的依赖
- 高质量数据更重要

## 模型

深度学习与经典方法的区别主要在于：前者关注功能强大的模型

## 目标函数

在机器学习中，我们需要定义对模型的优劣程度的度量，这个度量在大多数情况下时可优化的，这被称为目标函数，我们通常定义一个目标函数，并希望优化它到最小值。

因为越小越好，所以这些函数被称为损失函数

当任务在试图预测数值时，最常见的损失函数时平方误差（即预测值与实际值之差的平方）

- 欠拟合：模型对训练数据的拟合度不足，导致在测试数据上的表现不佳
- 过拟合：模型对训练数据的拟合度过高，导致在测试数据上的表现不佳

## 优化算法

当我们获得了高质量数据，一个模型，一个合适的损失函数，接下来需要一种算法，它能够搜索出最佳参数。

在深度学习中，大多数流行的优化算法通常基于一种基本方法：梯度下降。

梯度下降是一种基于梯度的优化算法，它通过重复地沿着损失函数的负梯度下降，来优化参数是的可以减少损失的方向上优化参数


# 各种机器学习问题

## 监督学习

监督学习擅长在“给定输入特征”的情况下预测标签

![监督学习](src/1-3.png)


### 回归

回归（Regression）是指预测目标变量（标签）为连续数值的学习任务。其核心目标是从标注数据中学习一个从输入特征空间到连续数值输出空间的映射函数 $f: \mathcal{X} \to \mathbb{R}$，从而对新输入的样本给出精准的数值估计。

### 分类

分类（Classification）是指预测目标变量（标签）为离散类别的学习任务。其核心目标是从标注数据中学习输入特征与类别标签之间的映射关系

不只是分类，还必须有序分类

### 序列学习

如果输入的样本之间没有任何关系，以上模型给肯能完美无缺，但是如果输入是连续的，模型就可能需要具有记忆功能，例如处理视频片段或文本序列。

- 标记和解析
- 自动语音识别
- 文本到语音
- 机器翻译

## 无监督学习

不含有目标的机器学习问题通常被称为无监督学习。

- 聚类
- 主成分分析
- 因果关系
- 生成对抗网络


## 强化学习

智能体在一系列的时间步骤上与环境交互，在每个特定时间点，智能体从环境接收一些观测，并且必须选择一个动作，然后通过某种机制（有时称为执行器）将其传输回环境，最后智能体从环境中获得奖励。循环往复。

强化学习的目标是产生一个好的策略
